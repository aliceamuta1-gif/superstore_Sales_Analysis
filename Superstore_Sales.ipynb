```python
import pandas as pd
df["OrderDate"]=pd.to_datetime(df["OrderDate"],errors='coerce')
df["Month"]=df["OrderDate"].dt.month_name()
```

```python
df.groupby("State")["Sales"].sum().sort_values(ascending=False).head()
```

```python
df.groupby("Category")["Profit"].sum().sort_values(ascending=False).head()
```

```python
df["Profit_Margin"]=(df["Profit"]/df["Sales"])*100
df.head()
```

```python
df.sort_values(["Sales","Profit_Margin"],ascending=[False,True]).head()
```

```python
df[df["Profit"]<0]
```

```python
df.sort_values("Profit_Margin",ascending=False).head()
```

```python
df.groupby("Category").agg({"Sales":["sum","mean"],"Profit":["sum","mean"],"Profit_Margin":["mean","max"]})
```

```python
df.groupby("State").agg({"Sales":["sum","mean"],"Profit":["sum","mean"],"Profit_Margin":["mean","max"]})
```

```python
Summary=df.groupby("State").agg({"Sales":["sum","mean"],"Profit":["sum","mean"],"Profit_Margin":["mean","max"]})
Summary.Columns=["Total_Sales","Average_Sales","Total_Profit","Average_Profit_Margin","max_Profit_Margin"]
Summary.head()
```

```python
State_Summary=df.groupby("State").agg({"Profit":"sum","Profit_Margin":"mean"})
State_Summary.sort_values(["Profit","Profit_Margin"],ascending=[False,True]).head()
```

```python
import matplotlib.pyplot as plt
df.groupby("Category")["Sales"].sum().plot(kind="bar")
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sale")
plt.show()
```

```python
df.groupby("State")["Profit_Margin"].mean().sort_values(ascending=False).head(10).plot(kind="bar")
plt.title("Top 10 States by Average profit margin")
plt.xlabel("State")
plt.ylabel("Average profit_Margin")
plt.show()
```

```python
df.groupby("Month")["Sales"].sum().sort_values(ascending=False).head(10).plot(kind="line")
plt.title("Total Sales by Months")
plt.xlabel("Month")
plt.ylabel("Total sales")
plt.show()
```
